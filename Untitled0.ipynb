```python
import collections
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(graph, root):
    visited = set()
    queue = collections.deque([root])
    visited.add(root)
    result = []

    while queue:
        vertex = queue.popleft()
        result.append(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return result

bfs_order = bfs(graph, 'A')
print(f"BFS Traversal starting from node A: {bfs_order}")

# Visualize with matplotlib
G = nx.DiGraph(graph)
pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_color='lightgreen', node_size=2000, font_size=15, arrows=True)
plt.title("Graph Representation for BFS")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

def dfs(graph, node, visited=None):
    if visited is None:
        visited = []
    if node not in visited:
        visited.append(node)
        for neighbor in graph[node]:
            dfs(graph, neighbor, visited)
    return visited

dfs_order = dfs(graph, 'A')
print(f"DFS Traversal starting from node A: {dfs_order}")

# Visualize with matplotlib
G = nx.DiGraph(graph)
pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2000, font_size=15, arrows=True)
plt.title("Graph Representation for DFS")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# 4. Print both traversal orders
print(f"BFS Traversal Order: {bfs_order}")
print(f"DFS Traversal Order: {dfs_order}")

def draw_traversal_tree(graph, traversal_order, title):
    G = nx.DiGraph()
    for node, neighbors in graph.items():
        for neighbor in neighbors:
            G.add_edge(node, neighbor)

    # Use a hierarchical layout for a tree-like look
    pos = nx.nx_agraph.graphviz_layout(G, prog='dot') if 'pygraphviz' in globals() else nx.spring_layout(G, seed=42)

    # Mapping visit order to nodes
    order_map = {node: i + 1 for i, node in enumerate(traversal_order)}
    labels = {node: f"{node}\n({order_map[node]})" for node in G.nodes()}

    plt.figure(figsize=(8, 6))
    nx.draw(G, pos, with_labels=True, labels=labels, node_size=2000,
            node_color='skyblue', font_size=10, font_weight='bold',
            arrows=True, edge_color='gray')
    plt.title(title)
    plt.show()

# Draw BFS tree
draw_traversal_tree(graph, bfs_order, "BFS Traversal Order (Node Name and Visit Index)")

# Draw DFS tree
draw_traversal_tree(graph, dfs_order, "DFS Traversal Order (Node Name and Visit Index)")
```

```python
from collections import deque

def water_jug_bfs(cap_a, cap_b, target):
    # (amount_a, amount_b)
    initial_state = (0, 0)
    queue = deque([(initial_state, [])])
    visited = set([initial_state])

    while queue:
        (a, b), path = queue.popleft()

        if a == target or b == target:
            return path + [(a, b)]

        # Possible next states
        states = [
            ((cap_a, b), "Fill Jug A"),
            ((a, cap_b), "Fill Jug B"),
            ((0, b), "Empty Jug A"),
            ((a, 0), "Empty Jug B"),
            ((max(0, a - (cap_b - b)), min(cap_b, b + a)), "Pour A -> B"),
            ((min(cap_a, a + b), max(0, b - (cap_a - a))), "Pour B -> A")
        ]

        for state, action in states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b, action)]))

    return None

# Solve for 2 litres
solution = water_jug_bfs(4, 3, 2)

if solution:
    print("Step-by-step solution to reach 2L:")
    for step in solution:
        print(step)
else:
    print("No solution found.")
```

```python
from collections import deque
import matplotlib.pyplot as plt

def water_jug_bfs(cap_a, cap_b, target):
    initial_state = (0, 0)
    queue = deque([(initial_state, [])])
    visited = {initial_state}

    while queue:
        (a, b), path = queue.popleft()

        if a == target or b == target:
            return path + [(a, b)]

        moves = [
            ((cap_a, b), "Fill A"),
            ((a, cap_b), "Fill B"),
            ((0, b), "Empty A"),
            ((a, 0), "Empty B"),
            ((max(0, a - (cap_b - b)), min(cap_b, b + a)), "A->B"),
            ((min(cap_a, a + b), max(0, b - (cap_a - a))), "B->A")
        ]

        for state, action in moves:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b, action)]))
    return None

result = water_jug_bfs(4, 3, 2)

if result:
    print("BFS Solution found. Plotting steps...")
    steps = len(result)
    fig, axes = plt.subplots(1, steps, figsize=(steps * 2.5, 4))
    if steps == 1: axes = [axes]

    for i, step in enumerate(result):
        a, b = (step[0], step[1]) if len(step) > 2 else step
        label = step[2] if len(step) > 2 else "Goal"

        axes[i].bar(['Jug A', 'Jug B'], [a, b], color=['blue', 'cyan'])
        axes[i].set_ylim(0, max(4, 3) + 1)
        axes[i].set_title(f"Step {i}: {label}\n({a}L, {b}L)")

    plt.tight_layout()
    plt.show()
else:
    print("No solution found.")
```

```python
import matplotlib.pyplot as plt

def visualize_jugs(solution):
    if not solution:
        print("No solution to visualize.")
        return

    steps = len(solution)
    fig, axes = plt.subplots(1, steps, figsize=(steps * 3, 5))

    if steps == 1:
        axes = [axes]

    for i, step in enumerate(solution):
        # Handle different tuple lengths in the path
        if len(step) == 3:
            a, b, label = step
        else:
            a, b = step
            label = "Goal"

        ax = axes[i]
        # Draw Jug A
        ax.bar(['Jug A'], [a], color='blue', alpha=0.6, width=0.6)
        ax.set_ylim(0, 4)
        # Draw Jug B
        ax.bar(['Jug B'], [b], color='cyan', alpha=0.6, width=0.6)
        ax.set_ylim(0, 4)

        ax.set_title(f"Step {i}\n{label}")
        ax.set_ylabel("Liters")

    plt.tight_layout()
    plt.show()

# Visualize the result from the previous BFS execution
visualize_jugs(result)
```

```python
import matplotlib.pyplot as plt
from collections import deque

def solve_water_jug(cap_a, cap_b, target):
    # BFS initialization: ((current_a, current_b), path_taken)
    queue = deque([((0, 0), [])])
    visited = set([(0, 0)])

    while queue:
        (a, b), path = queue.popleft()

        # Goal check
        if a == target or b == target:
            return path + [(a, b, "Goal Reached")]

        # Define all possible moves
        moves = [
            ((cap_a, b), "Fill Jug A"),
            ((a, cap_b), "Fill Jug B"),
            ((0, b), "Empty Jug A"),
            ((a, 0), "Empty Jug B"),
            # Pour A -> B: pour until A is empty or B is full
            ((max(0, a - (cap_b - b)), min(cap_b, b + a)), "Pour A -> B"),
            # Pour B -> A: pour until B is empty or A is full
            ((min(cap_a, a + b), max(0, b - (cap_a - a))), "Pour B -> A")
        ]

        for state, action in moves:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b, action)]))
    return None

# Parameters
CAP_A, CAP_B, TARGET = 4, 3, 2
solution_path = solve_water_jug(CAP_A, CAP_B, TARGET)

# Visualization
if solution_path:
    n_steps = len(solution_path)
    fig, axes = plt.subplots(1, n_steps, figsize=(n_steps * 3, 5))
    if n_steps == 1: axes = [axes]

    for i, (a, b, action) in enumerate(solution_path):
        ax = axes[i]
        ax.bar(['Jug A (4L)', 'Jug B (3L)'], [a, b], color=['dodgerblue', 'skyblue'])
        ax.set_ylim(0, max(CAP_A, CAP_B) + 0.5)
        ax.set_title(f"Step {i}: {action}\n({a}L, {b}L)")
        ax.set_ylabel("Litres")

    plt.tight_layout()
    plt.show()
else:
    print("No solution found.")
```
